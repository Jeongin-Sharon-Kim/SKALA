```python
# ==========================================================
# TO DO 3. Example Prompt 작성
# ==========================================================
#
# ChatPromptTemplate를 이용하여
# example_prompt를 작성하세요.
#
# Prompt는
# Human Message
#
# AI Message
#
# 순서로 구성합니다.
#
# 작성한 Prompt는
#
# example_prompt 변수에 저장하세요.
#
# ==========================================================
from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate


market_example_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "human",
            """
다음 시장을 조사하세요.

시장:
{market}

다음 항목을 포함하여 분석하세요.

1. 시장 규모
2. 주요 경쟁사
3. 시장 전망
"""
        ),
        (
            "ai",
            """
{analysis}
"""
        )
    ]
)


print("\n========== 시장 분석 결과 ==========\n")

print(response.content)
```

```python
# ==========================================================
# TO DO 4. Few-shot Prompt 생성
# ==========================================================
#
# FewShotChatMessagePromptTemplate를 이용하여
# Few-shot Prompt를 생성하세요.
#
# 다음 변수를 연결합니다.
#
# • examples
# • example_prompt
#
# 생성한 Prompt는
#
# few_shot_prompt 변수에 저장하세요.
#
# ==========================================================


from langchain_core.prompts import FewShotChatMessagePromptTemplate


market_few_shot_prompt = FewShotChatMessagePromptTemplate(
    examples=examples,
    example_prompt=market_example_prompt
)
```

```python
# ==========================================================
# TO DO 5. 최종 Prompt 작성
# ==========================================================
#
# ChatPromptTemplate를 이용하여
# 최종 Prompt를 작성하세요.
#
# 다음 순서로 구성합니다.
#
# ① System Prompt
#
# "당신은 Gartner의 Senior Research Analyst입니다."
#
# ② Few-shot Prompt
#
# ③ Human Prompt
#
# "{input}"
#
# 작성한 Prompt는
#
# final_prompt 변수에 저장하세요.
#
# ==========================================================


final_prompt = ChatPromptTemplate.from_messages(
    [
        (
            "system",
            """
당신은 Gartner의 Senior Research Analyst입니다.

제공된 시장조사 사례의 구성과 분석 방식을 참고하여
요청받은 시장을 사업 관점에서 분석하세요.

확인되지 않은 구체적인 수치를 임의로 생성하지 말고,
불확실한 내용은 전망 또는 추정임을 구분하여 작성하세요.
"""
        ),

        market_few_shot_prompt,

        (
            "human",
            "{input}"
        )
    ]
)
```

```python
# ==========================================================
# TO DO 6. Prompt 실행
# ==========================================================
#
# 다음 질문을 실행하세요.
#
# "생성형 AI 교육 시장을 조사하세요."
#
# 실행 결과를 출력하세요.
#
# ==========================================================


prompt_value = final_prompt.invoke(
    {
        "input": "생성형 AI 교육 시장을 조사하세요."
    }
)

response = llm.invoke(prompt_value)

print(response.content)
```

```python
# ==========================================================
# TO DO 7. StrOutputParser 생성
# ==========================================================
#
# StrOutputParser 객체를 생성하세요.
#
# 생성한 객체를
#
# parser 변수에 저장하세요.
#
# ==========================================================
from langchain_core.output_parsers import StrOutputParser


parser = StrOutputParser()
```

```python
# ==========================================================
# TO DO 8. LCEL Chain 구현
# ==========================================================
#
# LCEL을 이용하여
#
# Prompt
#
# ↓
#
# ChatOpenAI
#
# ↓
#
# StrOutputParser
#
# 순서의 Chain을 작성하세요.
#
# 생성한 Chain은
#
# chain 변수에 저장하세요.
#
# ==========================================================

chain = (
    final_prompt
    |
    llm
    |
    parser
)
```

```python
# ==========================================================
# TO DO 20 정답
# 최종 프로젝트 결과 통합 및 실무 활용 분석
# ==========================================================


# ==========================================================
# 1. 라이브러리 Import
# ==========================================================

from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser



# ==========================================================
# 3. 최종 프로젝트 보고서 Prompt 작성
# ==========================================================


project_report_prompt = ChatPromptTemplate.from_messages(
    [

        (
            "system",
            """
당신은 AI 기반 업무 자동화 프로젝트를 평가하는
최종 프로젝트 분석 전문가입니다.

지금까지 수행한 시장조사,
사업 전략 수립,
의사결정 결과를 종합하여
최종 프로젝트 보고서를 작성하세요.


반드시 다음 내용을 포함하세요.


# AI 기반 시장조사 및 사업 전략 프로젝트 결과 보고서


## 1. 프로젝트 개요

- 프로젝트 목적
- 적용 기술


## 2. Prompt Engineering 적용 결과

- Role Prompting
- Few-shot Prompt
- Output Format Control


## 3. LangChain LCEL Pipeline 구성

- Prompt
- LLM
- Output Parser
- Chain 구조


## 4. 최종 시장 분석 결과

- 시장 기회
- 성장 가능성


## 5. 사업 전략 결과

- 고객 전략
- 서비스 전략
- 차별화 전략


## 6. 최종 의사결정

- 시장 진입 여부
- 판단 근거


## 7. 실무 활용 방안

- 기업 업무 적용 방법
- 자동화 가능 업무


## 8. 개선 방향

- 추가 데이터 활용
- Agent 확장 방향


보고서는 실제 기업 경영진에게
제출 가능한 수준으로 작성하세요.
"""
        ),


        (
            "human",
            """
다음 프로젝트 결과를 기반으로
최종 보고서를 작성하세요.


[시장조사 결과]

{market_result}



[사업 전략 결과]

{strategy_result}



[의사결정 결과]

{decision_result}



[평가 결과]

{evaluation_result}

"""
        )

    ]
)



# ==========================================================
# 4. Output Parser 생성
# ==========================================================


parser = StrOutputParser()



# ==========================================================
# 5. 최종 Report Chain 구성
# ==========================================================


final_report_chain = (
    project_report_prompt
    |
    llm
    |
    parser
)



# ==========================================================
# 6. 프로젝트 결과 데이터 입력
# ==========================================================


project_input = {


    "market_result":
    """
생성형 AI 교육 서비스 시장 분석 결과:

- 생성형 AI 활용 증가
- 기업 AI 교육 수요 확대
- B2B 시장 성장 가능성 확인
- 실무 중심 교육 서비스 기회 존재
""",



    "strategy_result":
    """
사업 전략 결과:

- 기업 맞춤형 AI 교육 서비스 제공
- 직무별 생성형 AI 활용 과정 개발
- 프로젝트 기반 실습 교육 차별화
- 초기 기업 고객 확보 후 시장 확대
""",



    "decision_result":
    """
최종 의사결정:

시장 진입 추천

판단 근거:

- 높은 시장 성장 가능성
- 명확한 고객 요구
- 차별화된 서비스 제공 가능

주의 사항:

- 경쟁 심화
- 지속적인 콘텐츠 업데이트 필요
""",



    "evaluation_result":
    """
평가 결과:

Prompt Engineering 적용으로
AI 응답 품질과 일관성이 향상됨.

Few-shot Prompt 적용으로
보고서 형태 유지 가능.

LCEL Pipeline 구성으로
시장조사부터 의사결정까지 자동화 가능.
"""

}



# ==========================================================
# 7. 최종 프로젝트 보고서 생성
# ==========================================================


final_report = final_report_chain.invoke(
    project_input
)



# ==========================================================
# 8. 최종 결과 출력
# ==========================================================


print("=" * 70)

print("Prompt Engineering 기반 AI 사업 전략 프로젝트 최종 보고서")

print("=" * 70)


print(final_report)



# ==========================================================
# 9. 프로젝트 완료 구조 출력
# ==========================================================


print(
"""
================================================================

최종 프로젝트 Architecture


                 사용자 요구사항
                       |
                       v

              Prompt Engineering

        +-----------------------------+
        |                             |
        v                             v

   Role Prompt                 Few-shot Prompt

        |                             |
        +-------------+---------------+

                      |
                      v

              LCEL Pipeline

        Prompt
          |
          v
       ChatOpenAI
          |
          v
    StrOutputParser

                      |
                      v

          AI 시장조사 결과 생성

                      |
                      v

          사업 전략 수립

                      |
                      v

          경영 의사결정 지원


================================================================


프로젝트 핵심 성과

1. Prompt 설계 능력 확보
2. Few-shot 기반 품질 개선 경험
3. LCEL 기반 AI Pipeline 구현
4. AI 기반 업무 자동화 구조 이해
5. 실제 기업 의사결정 지원 프로세스 구현


================================================================
"""
)
```
